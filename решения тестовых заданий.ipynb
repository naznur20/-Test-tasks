```python
import pandas as pd
from scipy import stats

df = pd.read_csv('Данные для тестового задания - Данные АБ тестов.csv')


for exp in df['experiment_num'].unique():
    sub = df[df['experiment_num'] == exp]
    test = sub[sub['experiment_group'] == 'test']['revenue']
    control = sub[sub['experiment_group'] == 'control']['revenue']
    
    stat, p = stats.mannwhitneyu(test, control, alternative='two-sided')
    print(f"Эксперимент {exp}: ARPU test={test.mean():.2f}, "
          f"ARPU control={control.mean():.2f}, p-value={p:.4f}")
```
